# **Structural Damage Classification Modeling**

```python
import numpy as np
import pandas as pd

data_dir = "../data"
test = pd.read_csv(f"{data_dir}/test.csv")
sample_sub = pd.read_csv(f"{data_dir}/sample_submission.csv")
test_probs = np.load("../models/test_probs.npy")
offsets = np.load("../models/offsets.npy")
```

```python
test_pred = (np.log(test_probs + 1e-9) + offsets).argmax(1) + 1

labels = pd.Series(test_pred, index=test["building_id"])
submission = pd.DataFrame({"building_id": sample_sub["building_id"],
                           "damage_grade": sample_sub["building_id"].map(labels)})
assert submission["damage_grade"].notna().all()
assert submission.shape == sample_sub.shape
submission.to_csv("../submissions/submission.csv", index=False)
submission.head()
```
